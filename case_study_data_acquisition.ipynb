```python
#Task 1

import pandas as pd

population = pd.read_csv("/content/drive/MyDrive/DSA/Dataset/DataAcquisitionCasestudy_B10/population.csv")

print("First 5 rows:")
display(population.head())

print("Shape:")
print(population.shape)

print("Data Types:")
print(population.dtypes)
```

```python
#Task 2

gdp= pd.read_excel("/content/drive/MyDrive/DSA/Dataset/DataAcquisitionCasestudy_B10/gdp.xlsx")

print("First 5 rows:")
display(gdp.head())

print("Columns Name:")
print(gdp.columns)

print("Data Types:")
print(gdp.dtypes)
```

```python
#Task 3

internet_users = pd.read_json("/content/drive/MyDrive/DSA/Dataset/DataAcquisitionCasestudy_B10/internet_users.json")

df = pd.DataFrame(internet_users)
print(df)

print("Internet Users Dataset:")

display(internet_users)

print("First 5 rows:")

display(internet_users.head())

print("Shape:")

print(internet_users.shape)

print("Data Types:")

print(internet_users.dtypes)
```

```python
#Task 4

literacy_rate = pd.read_xml("/content/drive/MyDrive/DSA/Dataset/DataAcquisitionCasestudy_B10/literacy_rate.xml")

print("First 5 rows:")

display(literacy_rate.head())
print("")

print("Shape:")

print(literacy_rate.shape)
print("")

print("Data Types:")

print(literacy_rate.dtypes)
```

```python
#Task 5

datasets = {
    "Population": population,
    "GDP": gdp,
    "Internet Users": internet_users,
    "Literacy Rate": literacy_rate
}

for name, data in datasets.items():

    display(name)

    print("Shape:")
    display(data.shape)
    print("")

    print("Column Names:")
    display(data.columns.tolist())
    print("")

    print("Data Types:")
    display(data.dtypes)
    print("")

    print("Missing Values:")
    display(data.isnull().sum())
    print("")

    print("Duplicate Records:")
    display(data.duplicated().sum())
    print("")

    print("-" * 50)
    print("")

```

```python
# Task 6

gdp.rename(columns={"Country": "country"}, inplace=True)

internet_users.rename(columns={"Country": "country"}, inplace=True)

literacy_rate.rename(columns={"name": "country"}, inplace=True)

print("Population:")
display(population.columns)
print("")

print("GDP:")
display(gdp.columns)
print("")

print("Internet Users:")
display(internet_users.columns)
print("")

print("Literacy Rate:")
display(literacy_rate.columns)
```

```python
# Task 7

population["population"] = pd.to_numeric(population["population"])
gdp["GDP"] = pd.to_numeric(gdp["GDP"])
internet_users["internet_users"] = pd.to_numeric(internet_users["internet_users"])
literacy_rate["literacy_rate"] = pd.to_numeric(literacy_rate["literacy_rate"])


print("Population Data Types:")
display(population.dtypes)

print("GDP Data Types:")
display(gdp.dtypes)

print("Internet Users Data Types:")
display(internet_users.dtypes)

print("Literacy Rate Data Types:")
display(literacy_rate.dtypes)
```

```python
# Task 8

merged_data = population.merge(
    gdp,
    on="country",
    how="inner"
)

merged_data = merged_data.merge(
    internet_users,
    on="country",
    how="inner"
)

merged_data = merged_data.merge(
    literacy_rate,
    on="country",
    how="inner"
)

display(merged_data)
```

```python
# Task 9

merged_data["Internet Penetration (%)"] = (
    merged_data["internet_users"] / merged_data["population"] * 100
).round(2)

merged_data["GDP Per Capita"] = (
    merged_data["GDP"] / merged_data["population"]
).round(2)

display(merged_data)
```

```python
# Task 10

import sqlite3

connection = sqlite3.connect("country_statistics.db")

merged_data.to_sql(
    "country_statistics",
    connection,
    if_exists="replace",
    index=False
)

display("Database and table created successfully.")
```

```python
# Task 11

query = "SELECT * FROM country_statistics LIMIT 5"

first_five = pd.read_sql_query(query, connection)

display(first_five)
```
